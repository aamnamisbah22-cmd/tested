```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("tested.csv")
data.columns = data.columns.str.strip()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x=data["Age"], y=data["Fare"])
plt.title("Scatter Plot: Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.violinplot(y=data["Age"])
plt.title("Violin Plot of Age")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.violinplot(y=data["Fare"])
plt.title("Violin Plot of Fare")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.kdeplot(data["Age"].dropna(), fill=True)
plt.title("Density Plot of Age")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.kdeplot(data["Fare"].dropna(), fill=True)
plt.title("Density Plot of Fare")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.histplot(data["Age"], bins=20, kde=True)
plt.title("Histogram of Age")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.histplot(data["Fare"], bins=20, kde=True)
plt.title("Histogram of Fare")
plt.show()
```

```python
cols = ['PassengerId', 'Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

sns.pairplot(data[cols])
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(data[['PassengerId','Survived','Pclass','Age','SibSp','Parch','Fare']].corr(),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
